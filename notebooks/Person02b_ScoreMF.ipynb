```python
import re, subprocess

log = subprocess.run(["git","log","--all","-p","--no-color"],
                     capture_output=True, text=True, errors="ignore").stdout

# Walk the diff, track which file each line belongs to, skip data files
skip = (".csv", ".png", ".json", ".md", ".txt")
cur, hits = "", []
for line in log.splitlines():
    if line.startswith("diff --git"):
        cur = line.split(" b/")[-1]
        continue
    if cur.endswith(skip) or not line.startswith(("+", "-")):
        continue
    keyish = re.search(r"(api[_-]?key|token|secret|password|bearer)", line, re.I)
    hexval = re.search(r"\b[0-9a-fA-F]{32}\b", line)
    quoted = re.search(r"(key|token|secret|password)\w*\s*[=:]\s*[\"'][A-Za-z0-9_\-]{16,}[\"']", line, re.I)
    if (keyish and hexval) or quoted:
        hits.append((cur, line[:1] + line[1:60] + "..."))

print(len(hits), "suspicious lines")
for h in hits[:20]:
    print(h)
```
